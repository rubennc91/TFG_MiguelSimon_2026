## 1. Imports y configuración

Esta celda solo prepara FINN. También corrige el fallo que ya nos salió:

```text
AttributeError: 'str' object has no attribute 'value'
```

Ese fallo pasaba porque `default_mem_mode` se estaba pasando como texto. Aquí se pasa como `MemMode.INTERNAL_DECOUPLED`.


````python
from pathlib import Path
import os
import shutil
import subprocess

from finn.builder.build_dataflow import build_dataflow_cfg
from finn.builder.build_dataflow_config import (
    DataflowBuildConfig,
    DataflowOutputType,
    ShellFlowType,
)

# Según la versión de FINN, MemMode puede estar en un sitio u otro
try:
    from finn.builder.build_dataflow_config import MemMode
    print("MemMode importado desde finn.builder.build_dataflow_config")
except Exception:
    from finn.transformation.fpgadataflow.convert_to_hw_layers import MemMode
    print("MemMode importado desde finn.transformation.fpgadataflow.convert_to_hw_layers")

MODEL_TO_BUILD = Path("model.onnx")
OUTPUT_DIR = Path("output_final_mlp27k_w4a8_ft")
FINN_BUILD_DIR = Path("/tmp/FINN_build_mlp27k_w4a8_ft")

print("Modelo:", MODEL_TO_BUILD.resolve())
print("Output dir:", OUTPUT_DIR.resolve())
print("FINN_BUILD_DIR:", FINN_BUILD_DIR)
````

## 2. Comprobar que existe `model.onnx`

No seguimos si no existe, porque este notebook ya asume que el ONNX final está creado.


````python
if not MODEL_TO_BUILD.exists():
    raise FileNotFoundError(
        "No existe model.onnx en esta carpeta. Copia aquí el ONNX final cuantizado + fine-tuned antes de seguir."
    )

print("Existe model.onnx:", MODEL_TO_BUILD.exists())
print("Tamaño:", MODEL_TO_BUILD.stat().st_size / 1024, "KB")
````

## 3. Limpieza opcional de salidas anteriores

Déjalo en `True` si quieres empezar limpio. Esto borra:

```text
output_final_mlp27k_w4a8_ft
/tmp/FINN_build_mlp27k_w4a8_ft
```


````python
CLEAN_PREVIOUS_OUTPUTS = True

if CLEAN_PREVIOUS_OUTPUTS:
    if OUTPUT_DIR.exists():
        print("Borrando", OUTPUT_DIR)
        shutil.rmtree(OUTPUT_DIR)

    if FINN_BUILD_DIR.exists():
        print("Borrando", FINN_BUILD_DIR)
        shutil.rmtree(FINN_BUILD_DIR)

FINN_BUILD_DIR.mkdir(parents=True, exist_ok=True)
os.environ["FINN_BUILD_DIR"] = str(FINN_BUILD_DIR)

print("FINN_BUILD_DIR creado:", FINN_BUILD_DIR.exists())
````

## 4. Inspección rápida del ONNX

Esto no evalúa resultados. Solo enseña qué nodos tiene el modelo para saber si FINN está viendo un grafo razonable.


````python
from qonnx.core.modelwrapper import ModelWrapper
from qonnx.transformation.infer_shapes import InferShapes
from qonnx.transformation.infer_datatypes import InferDataTypes

model = ModelWrapper(str(MODEL_TO_BUILD))
model = model.transform(InferShapes())
model = model.transform(InferDataTypes())

print("Inputs:")
for inp in model.graph.input:
    name = inp.name
    try:
        dt = model.get_tensor_datatype(name)
    except Exception:
        dt = "?"
    print(" ", name, model.get_tensor_shape(name), dt)

print("\nOutputs:")
for out in model.graph.output:
    name = out.name
    try:
        dt = model.get_tensor_datatype(name)
    except Exception:
        dt = "?"
    print(" ", name, model.get_tensor_shape(name), dt)

print("\nNodos del modelo:")
for i, node in enumerate(model.graph.node):
    print(f"{i:03d}: {node.op_type:30s} {node.name}")
````

## 5. Lanzar FINN build

Esta versión intenta generar principalmente el `stitched_ip` y los informes de estimación.

No se incluyen al principio `RTLSIM_PERFORMANCE` ni `OOC_SYNTH` para evitar alargar mucho el flujo. Primero queremos conseguir el `stitched_ip`. Si esto termina bien, luego se puede lanzar una versión más completa.

Si parece que se queda parado, no abras varios notebooks a la vez. Mira el log desde terminal.


````python
cfg = DataflowBuildConfig(
    output_dir=str(OUTPUT_DIR),
    synth_clk_period_ns=10.0,
    board=None,
    shell_flow_type=ShellFlowType.NONE,
    generate_outputs=[
        DataflowOutputType.ESTIMATE_REPORTS,
        DataflowOutputType.STITCHED_IP,
    ],
    default_mem_mode=MemMode.INTERNAL_DECOUPLED,
    folding_config_file=None,
    target_fps=100,
    mvau_wwidth_max=36,
    verify_steps=[],
    save_intermediate_models=True,
)

print("Lanzando build_dataflow con", MODEL_TO_BUILD)
print("Output dir:", OUTPUT_DIR)
print("FINN_BUILD_DIR:", os.environ["FINN_BUILD_DIR"])

build_dataflow_cfg(str(MODEL_TO_BUILD), cfg)
````

## 6. Comprobar salida del build

Ejecuta esta celda cuando termine el build o si quieres ver qué se ha generado.


````python
print("Existe output dir:", OUTPUT_DIR.exists())

if OUTPUT_DIR.exists():
    print("Contenido de output dir:")
    for p in sorted(OUTPUT_DIR.iterdir()):
        print(" ", p)

stitched = OUTPUT_DIR / "stitched_ip"
print("\nExiste stitched_ip:", stitched.exists())

if stitched.exists():
    print("Contenido stitched_ip:")
    for p in sorted(stitched.iterdir()):
        print(" ", p)

print("\nIntermedios en FINN_BUILD_DIR:")
if FINN_BUILD_DIR.exists():
    onnx_files = sorted(FINN_BUILD_DIR.rglob("*.onnx"))
    print("ONNX intermedios:", len(onnx_files))
    for f in onnx_files[-20:]:
        print(" ", f)
else:
    print("No existe", FINN_BUILD_DIR)
````

## 7. Si falla o se queda en `step_convert_to_hls`

No hace falta tocar nada aquí salvo que falle. Esta celda lista los últimos ONNX intermedios y sus nodos.

Si el build se queda en:

```text
step_convert_to_hls
```

y la carpeta `/tmp/FINN_build_mlp27k_w4a8_ft` no crece, entonces seguramente FINN se ha quedado intentando transformar algún nodo.


````python
from pathlib import Path
from qonnx.core.modelwrapper import ModelWrapper

onnx_files = sorted(FINN_BUILD_DIR.rglob("*.onnx")) if FINN_BUILD_DIR.exists() else []
print("ONNX intermedios encontrados:", len(onnx_files))

for i, f in enumerate(onnx_files):
    print(i, f)

print("\nNodos de los últimos modelos:")
for f in onnx_files[-5:]:
    print("=" * 100)
    print(f)
    m = ModelWrapper(str(f))
    for node in m.graph.node:
        print(node.op_type, node.name)
````
